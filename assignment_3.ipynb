6. Explain the concept of generators in Python and how they are defined.

ANS: Generators in Python
A generator is a special type of iterator in Python that allows you to generate values on the fly instead of storing them all in memory at once. This makes generators memory-efficient, especially for large data sets.

How Are Generators Defined?
Generators are defined using:

Generator functions – functions that use the yield keyword.

Generator expressions – similar to list comprehensions but with parentheses.

1. Generator Function Example:
python
Copy
Edit
def count_up_to(n):
    count = 1
    while count <= n:
        yield count  # Pauses and returns a value
        count += 1

gen = count_up_to(3)
print(next(gen))  # 1
print(next(gen))  # 2
print(next(gen))  # 3
# next(gen) now would raise StopIteration
yield returns a value and pauses the function.

When next() is called again, it resumes from where it left off.

2. Generator Expression Example:
python
Copy
Edit
gen = (x*x for x in range(5))
print(next(gen))  # 0
print(next(gen))  # 1
Looks like a list comprehension, but uses () instead of [].

Key Benefits of Generators:
Memory-efficient: Do not store all values in memory.

Lazy evaluation: Values are generated only when needed.

Clean syntax for creating iterators.

Summary:
Generators are a concise way to write iterators.

Use yield in a function to turn it into a generator.

Ideal for large or infinite sequences where storing all values isn't practical.

Generators offer several benefits, especially when dealing with large data sets or streams of data.
1. Memory Efficiency
Generators don’t store all values in memory.

 2. Lazy Evaluation

This improves performance and avoids unnecessary computation.

10. What is the difference between `map()`, `reduce()`, and `filter()` functions in Python?

ANS: Difference Between map(), reduce(), and filter() in Python
These three functions—map(), filter(), and reduce()—are part of Python's functional programming tools, used to process iterables efficiently without writing explicit loops.


. map() – Transformation
Purpose: Applies a function to each item in an iterable and returns an iterator with the transformed items.

Syntax:

map(function, iterable)
Example:


numbers = [1, 2, 3]
squares = list(map(lambda x: x**2, numbers))  # [1, 4, 9]


2. filter() – Selection
Purpose: Applies a function that returns True or False to each item, and returns an iterator with only the items where the function returns True.

Syntax:

filter(function, iterable)
Example:

numbers = [1, 2, 3, 4]
evens = list(filter(lambda x: x % 2 == 0, numbers))  # [2, 4]

3. reduce() – Aggregation
Purpose: Applies a function cumulatively to the items in a sequence, reducing the iterable to a single value.

Note: reduce() is not a built-in function in Python 3. It must be imported from the functools module.

Syntax:

from functools import reduce
reduce(function, iterable)
Example:

from functools import reduce
numbers = [1, 2, 3, 4]
product = reduce(lambda x, y: x * y, numbers)  # 24

```python

#5. Create an iterator class in Python that generates the Fibonacci sequence up to a specified number of terms.

#ans:
class FibonacciIterator:
    def __init__(self, max_terms):
        self.max_terms = max_terms
        self.n1, self.n2 = 0, 1
        self.count = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.count >= self.max_terms:
            raise StopIteration
        if self.count == 0:
            self.count += 1
            return 0
        else:
            val = self.n2
            self.n1, self.n2 = self.n2, self.n1 + self.n2
            self.count += 1
            return val

# Testing
fib = FibonacciIterator(10)
for num in fib:
    print(num)
```

```python
# 6. Write a generator function in Python that yields the powers of 2 up to a given exponent.

def powers_of_two(n):
    for i in range(n + 1):
        yield 2 ** i

for power in powers_of_two(5):
    print(power)
```

```python
# 9. Write a Python program that uses `map()` to convert a list of temperatures from Celsius to Fahrenheit.

# List of temperatures in Celsius
celsius = [0, 20, 37, 100]

# Convert to Fahrenheit using map and a lambda function
fahrenheit = list(map(lambda c: (c * 9/5) + 32, celsius))

print(fahrenheit)
```

```python
# 10. Create a Python program that uses `filter()` to remove all the vowels from a given string.

# Given string
text = "Hello, how are you?"

# Function to check if a character is not a vowel
def not_vowel(char):
    return char.lower() not in 'aeiou'

# Use filter to remove vowels
filtered_chars = filter(not_vowel, text)

# Join the filtered characters back into a string
result = ''.join(filtered_chars)

print(result)
```
